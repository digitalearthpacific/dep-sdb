```python
from pystac import Item
from odc.stac import load
```

```python
item = Item.from_file("https://dep-public-staging.s3.us-west-2.amazonaws.com/dep_s2_sdb/0-1-1/057/042/2024/dep_s2_sdb_057_042_2024.stac-item.json")
data = load([item], chunks={})
data
```

```python
data = data.compute().squeeze()
```

```python
(data.pc_deep < 0.7).plot.imshow(size=10)
```

```python
(data.pc_deep < 0.8).plot.imshow(size=10)
```

```python
import folium
from ipyleaflet import basemaps

m = folium.Map(tiles=basemaps.Esri.WorldImagery)
m.fit_bounds(data.odc.map_bounds())

for var in data.data_vars:
    cmap = "Blues_r" if var in ("mean", "median", "depth") else "viridis"
    args = {
        "cmap": cmap,
    }
    if "pc" in var:
        args["vmin"] = 0
        args["vmax"] = 1
    data[var].odc.add_to(m, name=var, **args)

folium.LayerControl().add_to(m)

m
```
