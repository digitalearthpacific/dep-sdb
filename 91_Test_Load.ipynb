```python
from pystac import Item
from odc.stac import load, configure_s3_access
```

```python
configure_s3_access(cloud_defaults=True)

item = Item.from_file("https://dep-public-staging.s3.us-west-2.amazonaws.com/dep_s2_sdb/0-0-0/130/012/2024/dep_s2_sdb_130_012_2024.stac-item.json")

data = load([item], chunks={}).squeeze()
data
```

```python
data.odc.geobox.explore()
```
