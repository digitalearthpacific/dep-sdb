```python
from pystac import Item
from odc.stac import load, configure_s3_access
```

```python
configure_s3_access(cloud_defaults=True, aws_unsigned=True)

base = "https://dep-public-staging.s3.us-west-2.amazonaws.com/dep_s2_sdb"
a = "072"
b = "019"

item = Item.from_file(f"{base}/0-0-1/{a}/{b}/2024-07/2024-12/dep_s2_sdb_{a}_{b}_2024-07/2024-12.stac-item.json")

data = load([item], chunks={}).squeeze()
data.odc.geobox.explore()
```

```python
from ipyleaflet import basemaps

data["depth"].odc.explore(tiles=basemaps.Esri.WorldImagery)
```
